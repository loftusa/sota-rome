```python
from vllm import LLM, SamplingParams

kimi = "moonshotai/Moonlight-16B-A3B-Instruct"
sampling_params = SamplingParams(temperature=0.8, top_p=0.95)
llm = LLM(model=kimi, trust_remote_code=True)
```

```python
# apply chat template

from transformers import AutoTokenizer

tok = AutoTokenizer.from_pretrained(kimi, trust_remote_code=True)
```

```python
prompts = [
    "How does speculative decoding work?",
    "How does CPU offloading work in DeepSpeed?",
]
messages_list = [[{"role": "user", "content": prompt} for prompt in prompts]]
texts = tok.apply_chat_template(
    messages_list, tokenize=False, add_generation_prompt=True
)
for t in texts:
    print(t, "\n")
# outputs = llm.generate(texts)'\n'
```
